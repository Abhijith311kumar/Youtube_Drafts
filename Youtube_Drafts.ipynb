```python
from google.colab import userdata
import requests

url = "https://www.googleapis.com/youtube/v3/channels"
api_key = userdata.get('Youtube_API_Key')
params = {
    "part":"statistics",
    "forHandle":"sidemen",
    "key":api_key,
}

response = requests.get(url, params=params)
data = response.json()
print(data)
```

```python
subs = data['items'][0]['statistics']['subscriberCount']
print(subs)
```

```python
params = {
    "part":"contentDetails",
    "forHandle":"sidemen",
    "key":api_key,
}

response = requests.get(url, params=params)
data = response.json()
print(data)
```

```python
upload_id = data['items'][0]['contentDetails']['relatedPlaylists']['uploads']
print(upload_id)
```

```python
url = "https://www.googleapis.com/youtube/v3/videos"
params = {
    "part":"statistics",
    "id":"fwqEB2HdY88",
    "key":api_key,
}
response = requests.get(url, params=params)
data = response.json()
print(data)
```

Likes to views ratio for the Murder mystery video

```python
views = data['items'][0]['statistics']['viewCount']
likes = data['items'][0]['statistics']['likeCount']
like_ratio = (int(likes)/int(views))*100
print(f"Like ratio: {like_ratio:.2f}%")
```
